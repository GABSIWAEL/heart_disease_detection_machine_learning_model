```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('heart_disease_data.csv')
```

#1: Defective Heart
#0 : Healthy heart

```python
X = df.drop(columns='target' ,axis=1)    #remove the target from the data
```

```python
X
```

```python
Y=df['target']
```

```python
Y
```

SPLIT THE DATASET INTO TRAIN TEST DATA

```python
X_train , X_Test , Y_train , Y_test = train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=23)
```

```python
print(X_train.shape,X_Test.shape)
```

**MODEL TRAINING**

```python
model = LogisticRegression()
```

```python
#Train the model with training data
model.fit(X_train , Y_train)
```

```python
Y_pred = model.predict(X_Test)
```

```python
Y_train
```

```python
score = accuracy_score(Y_test,Y_pred)
```

```python
print ("the accuracy score is ", score )
```

**BUILDING A PREDICTIVE SYSTEM**

```python
input_data = (62,0,0,140,268,0,0,160,0,3.6,0,2,2)
#change the input_data into array

arr = np.array(input_data)
```

```python
print(arr)
```

```python
prediction= model.predict(arr.reshape(1,-1))
```

```python
print(prediction)
```

```python
if prediction == 0:
  print('You are Healthy')
else:
  print("Consult your doctor")
```

### Exporting the Trained Model

```python
import pickle

# Save the model to disk
filename = 'heart_disease_model.sav'
pickle.dump(model, open(filename, 'wb'))
print(f"Model saved successfully as '{filename}'!")
```

To verify that the model was exported properly, let's load it back and check if we can make a prediction with it.

```python
# Load the model from disk
loaded_model = pickle.load(open(filename, 'rb'))

# Test a prediction with the loaded model
test_prediction = loaded_model.predict(arr.reshape(1, -1))
print("Loaded model test prediction:", test_prediction)
```

```python
import pickle
import numpy as np

# 1. Load the exported model
loaded_model = pickle.load(open('heart_disease_model.sav', 'rb'))

# 2. Define a new person's data for prediction
# Format: (age, sex, cp, trestbps, chol, fbs, restecg, thalach, exang, oldpeak, slope, ca, thal)
new_input_data = (45, 1, 3, 110, 264, 0, 1, 132, 0, 1.2, 1, 0, 3)

# Convert to numpy array and reshape for prediction (1 sample, 13 features)
query_features = np.array(new_input_data).reshape(1, -1)

# 3. Make the prediction
new_prediction = loaded_model.predict(query_features)

print("Predicted class:", new_prediction[0])
if new_prediction[0] == 0:
  print('Healthy heart')
else:
  print('Defective heart (Consult your doctor)')
```
